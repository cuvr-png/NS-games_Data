# Análisis Exploratorio de Datos (EDA) - Nintendo Switch Games

Este notebook contiene la exploración inicial y la generación de prototipos de gráficos interactivos con **Plotly Express** para el dataset de juegos de Nintendo Switch.

```python
import pandas as pd
import plotly.express as px
```

```python
# Carga de datos con codificación adecuada (latin1)
df = pd.read_csv('Nintendo Switch Games.csv', encoding='latin1')
```

```python
# Limpieza de datos
df['Game'] = df['Game'].str.replace(r'\r?\nRead the review', '', regex=True)
df['Total_Shipped_M'] = df['Total Shipped'].str.replace('m', '', case=False).astype(float)

df.info()
df.head()
```

```python
# Histograma: Distribución de Puntajes de la Crítica (Critic Score)
fig_hist = px.histogram(df, x='Critic Score', title='Distribución de Critic Score')
fig_hist.show()
```

```python
# Gráfico de Dispersión: Critic Score vs User Score
fig_scatter1 = px.scatter(
    df.dropna(subset=['Critic Score', 'User Score']), 
    x='Critic Score', 
    y='User Score', 
    hover_name='Game',
    title='Critic Score vs User Score'
)
fig_scatter1.show()
```

```python
# Gráfico de Dispersión: Critic Score vs Ventas Totales en Millones
fig_scatter2 = px.scatter(
    df.dropna(subset=['Critic Score', 'Total_Shipped_M']), 
    x='Critic Score', 
    y='Total_Shipped_M', 
    hover_name='Game',
    title='Critic Score vs Ventas Totales (Millones)'
)
fig_scatter2.show()
```
